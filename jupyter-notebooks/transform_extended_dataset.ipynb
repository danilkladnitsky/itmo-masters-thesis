```python
import json
import re

# Input/output file paths
input_file = 'datasets/extended.json'
output_file = 'datasets/hsk1-extended-dataset.jsonl'

# Load input data
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# Process function
def transform_record(record):
    match = re.search(r'“(.*?)”', record.get("labeled_sentence", ""))
    if match:
        word = match.group(1)
        return {
            "word": word,
            "prompt": f"请用词语“{word}”造一个句子，只使用 HSK 1 级词汇，并且只写一句话。",
            "completion": record.get("original_sentence", "")
        }
    return None

# Transform records
transformed_data = [result for record in data if (result := transform_record(record))]

# Save as JSONL
with open(output_file, 'w', encoding='utf-8') as f:
    for item in transformed_data:
        f.write(json.dumps(item, ensure_ascii=False) + '\n')

print(f"Processed {len(transformed_data)} records. Output saved to {output_file}.")
```
